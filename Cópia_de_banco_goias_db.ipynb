```python


caminho_repasses = '/content/drive/MyDrive/Classroom/BUSINESS INTELLIGENCE IA2 2026_2/repasses_municipios_2014_2025'
arquivos_csv = sorted(glob.glob(os.path.join(caminho_repasses, '*.csv')))

print(f"Total de arquivos CSV encontrados: {len(arquivos_csv)}")

lista_dfs = []
for arquivo in arquivos_csv:
    try:
        # Tenta leitura padrão do portal Dados Abertos Goiás (ponto e vírgula + latin1)
        df_temp = pd.read_csv(arquivo, sep=';', encoding='latin1', low_memory=False)
        # Caso o arquivo venha separado por vírgula, relê com sep=','
        if df_temp.shape[1] == 1:
            df_temp = pd.read_csv(arquivo, sep=',', encoding='latin1', low_memory=False)
        lista_dfs.append(df_temp)
    except Exception as e:
        print(f"Erro ao ler {os.path.basename(arquivo)}: {e}")

data_repasses = pd.concat(lista_dfs, ignore_index=True)

print(f"Base 'data_repasses' consolidada com sucesso! Total de linhas: {len(data_repasses):,}")
print("Colunas encontradas:", list(data_repasses.columns))
data_repasses.head(3)
```

```python
import unicodedata
import numpy as np
import pandas as pd
import statsmodels.api as sm

def normalizar_texto(texto):
    """Remove acentos, espaços extras e padroniza em maiúsculas."""
    if pd.isna(texto):
        return ""
    texto_sem_acento = "".join(
        c for c in unicodedata.normalize("NFD", str(texto))
        if unicodedata.category(c) != "Mn"
    )
    return texto_sem_acento.upper().strip()

# --- TRATAMENTO DE SEGURANÇA SE OS DADOS DE REPASSES NÃO FOREM LOCALIZADOS ---
if 'data_repasses' not in locals() and 'data_repasses' not in globals():
    print("⚠️ 'data_repasses' não definido no notebook. Gerando dados de repasse simulados para viabilizar as análises...")
    # Gera dados de ICMS coerentes com os municípios que temos em 'df'
    municipios_unicos = df['municipio'].dropna().unique()
    np.random.seed(42)
    valores_icms = np.random.lognormal(mean=19, sigma=1.2, size=len(municipios_unicos))

    data_repasses = pd.DataFrame({
        'Nome do Município': municipios_unicos,
        'VALOR_BRUTO_ICMS': valores_icms
    })

# 1. Identificar coluna de município e valor de ICMS em data_repasses
col_mun_repasses = [
    c for c in data_repasses.columns
    if "MUNICIPIO" in c.upper() or "NOME" in c.upper()
][0]

data_repasses["municipio_chave"] = data_repasses[col_mun_repasses].apply(normalizar_texto)
df["municipio_chave"] = df["municipio"].apply(normalizar_texto)

colunas_icms = [
    c for c in data_repasses.columns
    if "CREDITADO_ICMS" in c.upper() or "VALOR_BRUTO_ICMS" in c.upper()
]
if colunas_icms:
    col_val = colunas_icms[0]
    if data_repasses[col_val].dtype == "object":
        data_repasses[col_val] = (
            data_repasses[col_val]
            .astype(str)
            .str.replace(".", "", regex=False)
            .str.replace(",", ".", regex=False)
        )
    data_repasses[col_val] = pd.to_numeric(data_repasses[col_val], errors="coerce").fillna(0.0)
else:
    col_val = data_repasses.select_dtypes(include=[np.number]).columns[-1]
    data_repasses[col_val] = pd.to_numeric(data_repasses[col_val], errors="coerce").fillna(0.0)

repasses_consolidado = (
    data_repasses.groupby("municipio_chave", as_index=False)[col_val]
    .sum()
    .rename(columns={col_val: "total_icms_repassado"})
)

# 2. Agregar auditorias por município (métricas reais e completas)
for col_num in ["qtd_constatacoes", "qtd_constatacoes_nao_conforme"]:
    if col_num in df.columns:
        df[col_num] = pd.to_numeric(df[col_num], errors="coerce").fillna(0)
    else:
        df[col_num] = 0

auditorias_consolidadas = df.groupby("municipio_chave", as_index=False).agg(
    total_auditorias=("numero_processo", "count"),
    total_constatacoes=("qtd_constatacoes", "sum"),
    total_nao_conformidades=("qtd_constatacoes_nao_conforme", "sum"),
)

# 3. Cruzamento (inner merge) e padronização final
df_completo = pd.merge(
    auditorias_consolidadas,
    repasses_consolidado,
    on="municipio_chave",
    how="inner",
)

df_completo["municipio"] = df_completo["municipio_chave"]
df_completo = df_completo.drop(columns=["municipio_chave"])

df_completo = (
    df_completo.dropna(subset=["municipio", "total_icms_repassado"])
    .drop_duplicates(subset=["municipio"])
    .sort_values("municipio")
    .reset_index(drop=True)
)

cols_numericas = [
    "total_auditorias",
    "total_constatacoes",
    "total_nao_conformidades",
    "total_icms_repassado",
]
for c in cols_numericas:
    df_completo[c] = pd.to_numeric(df_completo[c], errors="coerce").fillna(0.0)

# Aplica filtro caso municipios_selecionados esteja preenchido
municipios_selecionados = (
    municipios_selecionados if "municipios_selecionados" in locals() else []
)
df_filtrado = (
    df_completo[df_completo["municipio"].isin(municipios_selecionados)].copy()
    if municipios_selecionados
    else df_completo.copy()
)
df_filtrado_completo = df_filtrado.copy()

# 4. Regressão Linear Simples com statsmodels
col_x = "total_icms_repassado"
col_y = col_y if "col_y" in locals() else "total_constatacoes"
label_y = (
    label_y if "label_y" in locals() else "Total de Constatações em Auditorias"
)

X = df_filtrado[col_x]
Y = df_filtrado[col_y]

correlacao_r = X.corr(Y)
X_const = sm.add_constant(X)
modelo = sm.OLS(Y, X_const).fit()

intercepto_b = modelo.params["const"]
inclinacao_a = modelo.params[col_x]
r_quadrado = modelo.rsquared

print(f"Municípios Analisados: {len(df_filtrado)}")
print(f"Correlação de Pearson (r): {correlacao_r:.4f}")
print(f"Coef. Determinação (R²): {r_quadrado:.4f}")
print(f"Total ICMS na Amostra: R$ {X.sum():,.2f}")
print(
    f"\nEquação da Reta Ajustada: Y = {intercepto_b:.4f} + ({inclinacao_a:.8e}) * X"
)
```

```python

x_range = np.linspace(X.min(), X.max(), 100)
y_pred_line = intercepto_b + inclinacao_a * x_range

fig_reg = px.scatter(
    df_filtrado,
    x=col_x,
    y=col_y,
    hover_name="municipio",
    labels={col_x: "Total Repassado de ICMS (R$)", col_y: label_y},
    title=f"<b>Relação Linear:</b> Repasse ICMS (X) vs. {label_y} (Y)",
    template="plotly_dark"
)


fig_reg.update_traces(
    marker=dict(
        size=11,
        color='#1E88E5',
        opacity=0.85,
        line=dict(width=1.5, color='#90CAF9')
    ),
    name="Municípios",
    showlegend=True
)


fig_reg.add_trace(go.Scatter(
    x=x_range,
    y=y_pred_line,
    mode='lines',
    name='Reta de Regressão (OLS)',
    line=dict(color='#E53935', width=3)
))


fig_reg.update_layout(
    paper_bgcolor='#0B0E14',
    plot_bgcolor='#111622',
    font=dict(family="Arial, sans-serif", size=13, color="#E0E6ED"),
    title=dict(font=dict(size=18, color="#FFFFFF"), x=0.02),
    xaxis=dict(
        showgrid=True,
        gridcolor='rgba(255, 255, 255, 0.08)',
        zeroline=False
    ),
    yaxis=dict(
        showgrid=True,
        gridcolor='rgba(255, 255, 255, 0.08)',
        zeroline=False
    ),
    legend=dict(
        bgcolor='rgba(17, 22, 34, 0.8)',
        bordercolor='rgba(255, 255, 255, 0.15)',
        borderwidth=1,
        x=0.02,
        y=0.98
    ),
    hovermode="closest"
)

fig_reg.show()

```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import sqlite3
import statsmodels.api as sm
import plotly.express as px
import plotly.graph_objects as go

st.set_page_config(page_title="BI Fiscal e Saúde - Goiás", layout="wide")

st.title("📊 Inteligência Fiscal e Socioeconômica: ICMS vs. Auditorias de Saúde (GO)")
st.markdown("Análise de correlação e regressão linear simples entre repasses constitucionais de ICMS e fiscalizações em saúde nos municípios goianos.")

@st.cache_data
def carregar_dados_banco():
    conn = sqlite3.connect('projeto_bi.db')
    query = "SELECT * FROM indicadores_municipais"
    df = pd.read_sql_query(query, conn)
    conn.close()
    return df

df = carregar_dados_banco()

st.sidebar.header("⚙️ Configurações da Análise")

municipios_selecionados = st.sidebar.multiselect(
    "Filtrar Municípios (deixe vazio para todos):",
    options=sorted(df['municipio'].unique()),
    default=[]
)

df_filtrado = df[df['municipio'].isin(municipios_selecionados)] if municipios_selecionados else df.copy()

opcoes_y = {
    "Total de Constatações em Auditorias": "total_constatacoes",
    "Total de Não Conformidades": "total_nao_conformidades",
    "Quantidade de Auditorias Realizadas": "total_auditorias"
}
label_y = st.sidebar.selectbox("Selecione a Variável Resposta (Y):", list(opcoes_y.keys()))
col_y = opcoes_y[label_y]
col_x = "total_icms_repassado"

X = df_filtrado[col_x]
Y = df_filtrado[col_y]

correlacao_r = X.corr(Y)

X_const = sm.add_constant(X)
modelo = sm.OLS(Y, X_const).fit()

intercepto_b = modelo.params['const']
inclinacao_a = modelo.params[col_x]
r_quadrado = modelo.rsquared

col1, col2, col3, col4 = st.columns(4)
col1.metric("Municípios Analisados", len(df_filtrado))
col2.metric("Correlação de Pearson (r)", f"{correlacao_r:.4f}")
col3.metric("Coef. Determinação (R²)", f"{r_quadrado:.4f}")
col4.metric("Total ICMS na Amostra", f"R$ {X.sum():,.2f}")

st.info(f"**Equação da Reta Ajustada:** $\hat{{Y}} = {intercepto_b:.4f} + ({inclinacao_a:.8f}) \cdot X$")

tab1, tab2, tab3 = st.tabs(["📈 Regressão Linear", "📊 Análise Descritiva", "🔮 Simulador Preditivo"])

with tab1:
    st.subheader(f"Dispersão e Reta de Ajuste: Repasse ICMS (X) vs. {label_y} (Y)")

    x_range = np.linspace(X.min(), X.max(), 100)
    y_pred_line = intercepto_b + inclinacao_a * x_range

    fig_reg = px.scatter(
        df_filtrado, x=col_x, y=col_y, hover_name="municipio",
        labels={col_x: "Total Repassado de ICMS (R$)", col_y: label_y},
        title="Relação Linear entre Arrecadação de ICMS e Indicadores de Auditoria"
    )
    fig_reg.add_trace(go.Scatter(x=x_range, y=y_pred_line, mode='lines', name='Reta de Regressão (OLS)', line=dict(color='red', width=2)))
    st.plotly_chart(fig_reg, use_container_width=True)

with tab2:
    c1, c2 = st.columns(2)
    with c1:
        top10 = df_filtrado.nlargest(10, col_x)
        fig_bar = px.bar(
            top10, x='municipio', y=col_x,
            title="Top 10 Municípios por Volume de ICMS Repassado",
            labels={'municipio': 'Município', col_x: 'Total ICMS (R$)'}
        )
        st.plotly_chart(fig_bar, use_container_width=True)
    with c2:
        fig_box = px.box(
            df_filtrado, y=col_y, points="all", hover_name="municipio",
            title=f"Distribuição (Boxplot) - {label_y}",
            labels={col_y: label_y}
        )
        st.plotly_chart(fig_box, use_container_width=True)

with tab3:
    st.subheader("Simulador Preditivo em Tempo Real")
    st.markdown("Insira um valor hipotético de repasse de ICMS ($X$) para estimar o comportamento da variável resposta ($\hat{Y}$):")

    valor_x_input = st.number_input(
        "Valor de Repasse de ICMS (R$):",
        min_value=0.0,
        value=float(X.median()),
        step=100000.0,
        format="%.2f"
    )

    y_estimado = intercepto_b + (inclinacao_a * valor_x_input)

    st.success(f"Para um repasse de **R$ {valor_x_input:,.2f}**, o valor estimado de **{label_y}** ($\hat{{Y}}$) é de **{max(0, y_estimado):.2f}**.")

with st.expander("Visualizar Tabela Consolidada do Banco de Dados"):
    st.dataframe(df_filtrado)
```
